```python
import pandas as pd
from textblob import TextBlob
import matplotlib.pyplot as plt
```

```python
comments = [
    "Mumbai Indians played brilliantly today",
    "Chennai Super Kings gave an outstanding performance",
    "The batting was terrible and disappointing",
    "Excellent bowling by Jasprit Bumrah",
    "The match was boring and slow",
    "Virat Kohli played an amazing innings",
    "The team showed poor fielding",
    "Fantastic captaincy by MS Dhoni",
    "The match was exciting and thrilling",
    "Worst performance of the season"
]

df = pd.DataFrame(comments, columns=["Comment"])

df.head()
```

```python
def get_sentiment(text):

    polarity = TextBlob(text).sentiment.polarity

    if polarity > 0:
        return "Positive"
    elif polarity < 0:
        return "Negative"
    else:
        return "Neutral"
```

```python
df["Sentiment"] = df["Comment"].apply(get_sentiment)

df
```

```python
df["Sentiment"].value_counts()
```

```python
df["Sentiment"].value_counts().plot(
    kind="bar"
)

plt.title("IPL Sentiment Analysis")
plt.xlabel("Sentiment")
plt.ylabel("Count")

plt.show()
```

```python
df["Sentiment"].value_counts().plot(
    kind="pie",
    autopct="%1.1f%%"
)

plt.title("Sentiment Distribution")
plt.ylabel("")

plt.show()
```
