```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_style("whitegrid")
```

```python
df = pd.read_csv("../Task1_WebScraping/ipl_scraped_data.csv")
```

```python
plt.figure(figsize=(12,6))

df['winner'].value_counts().head(10).plot(kind='bar')

plt.title("Top 10 IPL Teams by Match Wins")
plt.xlabel("Team")
plt.ylabel("Number of Wins")

plt.xticks(rotation=45)

plt.tight_layout()

plt.show()
```

```python
plt.figure(figsize=(12,6))

df['toss_winner'].value_counts().head(10).plot(kind='bar')

plt.title("Top Teams Winning Toss")
plt.xlabel("Team")
plt.ylabel("Toss Wins")

plt.xticks(rotation=45)

plt.tight_layout()

plt.show()
```

```python
matches_per_season = df['season'].value_counts().sort_index()

plt.figure(figsize=(10,5))

plt.plot(matches_per_season.index,
         matches_per_season.values,
         marker='o')

plt.title("IPL Matches Per Season")
plt.xlabel("Season")
plt.ylabel("Matches")

plt.grid(True)

plt.show()
```

```python
plt.figure(figsize=(12,6))

df['player_of_match'].value_counts().head(10).plot(kind='bar')

plt.title("Top Player of the Match Award Winners")
plt.xlabel("Player")
plt.ylabel("Awards")

plt.xticks(rotation=45)

plt.tight_layout()

plt.show()
```

```python
plt.figure(figsize=(12,6))

df['venue'].value_counts().head(10).plot(kind='bar')

plt.title("Top IPL Venues")
plt.xlabel("Venue")
plt.ylabel("Matches Hosted")

plt.xticks(rotation=90)

plt.tight_layout()

plt.show()
```

```python
df['toss_decision'].value_counts().plot(
    kind='pie',
    autopct='%1.1f%%'
)

plt.title("Toss Decision Distribution")
plt.ylabel("")

plt.show()
```

```python
df['winner'].value_counts().head(5).plot(
    kind='pie',
    autopct='%1.1f%%'
)

plt.title("Top 5 IPL Teams by Wins")
plt.ylabel("")

plt.show()
```
